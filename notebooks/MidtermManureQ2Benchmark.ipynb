```python
import json
import pandas as pd
from IPython.display import Markdown, display

from src.midterm_benchmark import (
    DEFAULT_Q2_BENCHMARK_DIR,
    MIDTERM_Q2_REASONING_PROMPTS,
    MidtermBenchmarkConfig,
    load_benchmark_files,
    run_midterm_manure_q2_benchmark,
    write_midterm_outputs,
)
```

## Problem Statement And Reference Solution

```python
files = load_benchmark_files(DEFAULT_Q2_BENCHMARK_DIR)
reference = files["reference_solution"]

display(Markdown(files["problem_statement"]))

reference_summary = pd.DataFrame([
    {"metric": "objective_value", "value": reference["objective_value"]},
    {"metric": "demand_revenue", "value": reference["demand_revenue"]},
    {"metric": "transport_cost", "value": reference["transport_cost"]},
    {"metric": "supply_cost", "value": reference["supply_cost"]},
    {"metric": "supply_contribution", "value": reference["supply_contribution"]},
])
display(reference_summary)
display(reference)
```

## Canonical And Paraphrased Prompt Runs

```python
USE_LLM = bool(os.environ.get("GEMINI_API_KEY"))

config = MidtermBenchmarkConfig(
    prompt_ids=("canonical", "paraphrased"),
    use_llm=USE_LLM,
    use_llm_for_reasoning=USE_LLM,
    fallback_to_reference_fixture=True,
    attempt_solve=True,
    run_reasoning=False,
)
report = run_midterm_manure_q2_benchmark(config=config)

display(pd.DataFrame([report["metadata"]]))
display(report["tables"]["case_summary"])
display(report["tables"]["solve_accuracy"])
display(report["tables"]["interpretation_errors"])
display(report["tables"]["semantic_count_metrics"])
display(report["tables"]["parameter_multiset_metrics"])
display(report["tables"]["topology_metrics"])
display(report["tables"]["route_economics_metrics"])
display(report["tables"]["route_association_metrics"])
display(report["tables"]["transport_link_attribute_metrics"])
display(report["tables"]["solver_aggregate_metrics"])
display(report["tables"]["balance_residual_metrics"])
display(report["tables"]["formulation_completeness_metrics"])
display(report["tables"]["solve_correctness_metrics"])
display(report["tables"]["reasoning_readiness_metrics"])
display(report["tables"]["alias_resolution_diagnostics"])
display(report["tables"]["interpretation_metadata"])
```

```python
canonical_case = report["cases"][0]
display(canonical_case["validation_result"])
display(canonical_case["solution_checks"]["actual_components"])
```

## DNR Policy Diagnostics

```python
negative_bid_rows = report["tables"]["topology_metrics"].query(
    "metric == 'negative_bid_detection'"
)
route_rows = report["tables"]["route_economics_metrics"]
aggregate_rows = report["tables"]["solver_aggregate_metrics"]
route_economics_table = pd.DataFrame([
    {"route": "Eau Claire -> Menomonie", "consumer_bid": -0.5, "supplier_bid": 0.0, "transport_cost": 0.1, "route_net_value": -0.6},
    {"route": "Eau Claire -> Black River Falls", "consumer_bid": 1.5, "supplier_bid": 0.0, "transport_cost": 0.2, "route_net_value": 1.3},
])
q2_q3_comparison_table = pd.DataFrame([
    {"metric": "supplier_bid", "q2": 0.0, "q3": -0.7},
    {"metric": "accepted_supply", "q2": 500, "q3": 1000},
    {"metric": "Menomonie_flow", "q2": 0, "q3": 500},
    {"metric": "objective_value", "q2": 650, "q3": 1050},
])

display(negative_bid_rows)
display(route_rows)
display(route_economics_table)
display(q2_q3_comparison_table)
display(aggregate_rows[aggregate_rows["metric"].isin([
    "objective_value",
    "demand_revenue",
    "transport_cost",
    "total_accepted_supply",
    "total_accepted_demand",
    "active_transport_routes",
    "sorted_active_flow_values",
])])
```

## Required Reasoning Prompts

```python
display(pd.DataFrame(MIDTERM_Q2_REASONING_PROMPTS)[["id", "label", "prompt"]])

reasoning_config = MidtermBenchmarkConfig(
    prompt_ids=("canonical",),
    use_llm=USE_LLM,
    use_llm_for_reasoning=USE_LLM,
    fallback_to_reference_fixture=True,
    attempt_solve=True,
    run_reasoning=True,
)
reasoning_report = run_midterm_manure_q2_benchmark(config=reasoning_config)
display(reasoning_report["tables"]["reasoning_prompt_success"])

for row in reasoning_report["cases"][0]["reasoning_results"]:
    display(Markdown(f"### {row['prompt_label']}\n\n{row['response_preview']}"))
```
